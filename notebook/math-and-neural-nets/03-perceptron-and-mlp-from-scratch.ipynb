Open-source NumPy implementation of the 1957 Perceptron and a 2-layer MLP solving XOR.


```python
import numpy as np

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
y_and = np.array([0, 0, 0, 1], dtype=float)
y_xor = np.array([0, 1, 1, 0], dtype=float)

def train_perceptron(X, y, epochs=20, lr=0.1):
    w, b = np.zeros(X.shape[1]), 0.0
    for _ in range(epochs):
        for xi, target in zip(X, y):
            pred = 1.0 if (np.dot(xi, w) + b) >= 0 else 0.0
            err = target - pred
            w += lr * err * xi
            b += lr * err
    preds = [1 if (np.dot(xi, w) + b) >= 0 else 0 for xi in X]
    return preds, w, b

print('Perceptron AND:', train_perceptron(X, y_and)[0])
print('Perceptron XOR (fails):', train_perceptron(X, y_xor)[0])
```

```python
# 2-Layer ReLU MLP solving XOR
W1 = np.array([[1.0, 1.0], [1.0, 1.0]])
b1 = np.array([-0.5, -1.5])
W2 = np.array([[1.0], [-2.0]])
b2 = np.array([-0.2])

hidden = np.maximum(0.0, X @ W1 + b1)
out = (hidden @ W2 + b2).ravel()
print('2-Layer MLP XOR preds:', (out > 0).astype(int).tolist())
```
