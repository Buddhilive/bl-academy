# Lesson 4: Backpropagation, Activation Functions & Loss Functions

**Runtime Tier**: Google Colab Free Tier **CPU** (`numpy`)

Build a modular `Linear`, `ReLU`, and `Sequential` mini-framework with full backward passes.


```python
import numpy as np

class Linear:
    def __init__(self, in_f, out_f, rng):
        self.W = rng.normal(0.0, np.sqrt(2.0 / in_f), (in_f, out_f))
        self.b = np.zeros((1, out_f))

    def forward(self, x):
        self.x = x
        return x @ self.W + self.b

    def backward(self, grad_out, lr):
        dW = self.x.T @ grad_out
        db = grad_out.sum(axis=0, keepdims=True)
        dx = grad_out @ self.W.T
        self.W -= lr * dW
        self.b -= lr * db
        return dx

class ReLU:
    def forward(self, x):
        self.mask = x > 0
        return x * self.mask

    def backward(self, grad_out, lr=0.0):
        return grad_out * self.mask
```

```python
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

rng = np.random.default_rng(7)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
y = np.array([[0], [1], [1], [0]], dtype=float)

l1, act, l2 = Linear(2, 8, rng), ReLU(), Linear(8, 1, rng)
lr = 0.5
for epoch in range(2000):
    logits = l2.forward(act.forward(l1.forward(X)))
    y_hat = sigmoid(logits)
    loss = -np.mean(y * np.log(y_hat + 1e-8) + (1 - y) * np.log(1 - y_hat + 1e-8))
    grad = (y_hat - y) / len(X)
    grad = l2.backward(grad, lr)
    grad = act.backward(grad)
    l1.backward(grad, lr)

print('Final loss:', round(float(loss), 4))
print('Predictions:', np.round(y_hat.ravel(), 3).tolist())
```
