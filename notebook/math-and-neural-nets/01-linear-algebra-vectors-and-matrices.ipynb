```python
import numpy as np
import matplotlib.pyplot as plt

def cosine_similarity(a: np.ndarray, b: np.ndarray) -> float:
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))

v1 = np.array([1.0, 2.0, 0.5])
v2 = np.array([2.0, 3.5, 1.0])
v3 = np.array([-1.0, 0.5, 0.0])
print('cos(v1, v2):', round(cosine_similarity(v1, v2), 4))
print('cos(v1, v3):', round(cosine_similarity(v1, v3), 4))
```

```python
# Linear Transformations: Rotation + Shear
theta = np.pi / 4
R = np.array([[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]])
S = np.array([[1.0, 1.5], [0.0, 1.0]])
print('det(R):', round(float(np.linalg.det(R)), 4))
print('det(S):', round(float(np.linalg.det(S)), 4))

grid = np.array([[0, 0], [1, 0], [1, 1], [0, 1], [0, 0]], dtype=float)
rot_grid = grid @ R.T
shear_grid = grid @ S.T
print('Rotated square corners:\n', np.round(rot_grid, 3))
print('Sheared square corners:\n', np.round(shear_grid, 3))
```

```python
# Batched Linear Layer Forward Pass
rng = np.random.default_rng(42)
X = rng.normal(size=(32, 128))
W = rng.normal(scale=0.1, size=(64, 128))
b = np.zeros((1, 64))
Y = X @ W.T + b
print('Output shape:', Y.shape)
```
