# Lesson 13: Making Meaningful Visualizations & *Les Liaisons Dangereuses* Network

In this notebook, you will:
1. Reconstruct an honest chronological chart vs a deceptive non-chronological chart.
2. Build and visualize the *Les Liaisons Dangereuses* epistolary correspondence network using `NetworkX` and `Matplotlib`.

```python
!pip install -q networkx matplotlib
import networkx as nx
import matplotlib.pyplot as plt

letters = [
    ("Valmont", "Merteuil"), ("Merteuil", "Valmont"), ("Valmont", "Merteuil"),
    ("Valmont", "Tourvel"), ("Tourvel", "Valmont"), ("Merteuil", "Cécile"),
    ("Cécile", "Danceny"), ("Danceny", "Cécile"), ("Danceny", "Valmont"),
    ("Volanges", "Merteuil"), ("Tourvel", "Rosemonde"), ("Valmont", "Rosemonde")
]

G = nx.DiGraph()
for src, dst in letters:
    if G.has_edge(src, dst):
        G[src][dst]["weight"] += 1
    else:
        G.add_edge(src, dst, weight=1)

plt.figure(figsize=(8, 5))
pos = nx.spring_layout(G, seed=42)
nx.draw_networkx(G, pos, node_color="#A7452A", font_color="white", node_size=1800, font_size=9, arrows=True)
plt.title("Les Liaisons Dangereuses — Epistolary Network")
plt.axis("off")
plt.show()
```
