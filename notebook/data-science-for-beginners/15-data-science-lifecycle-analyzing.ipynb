# Lesson 15: Data Science Lifecycle — Exploratory Data Analysis (EDA)

In this notebook, you will:
1. Profile an Email Spam word-count dataset using `df.describe()`.
2. Draw random samples (`df.sample()`), run targeted queries (`df.query()`), and audit missingness (`df.isna().sum()`).

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(15)
n = 200
spam = rng.integers(0, 2, size=n)
df = pd.DataFrame({
    "the": rng.poisson(5, size=n),
    "free": np.where(spam == 1, rng.poisson(3.2, size=n), rng.poisson(0.3, size=n)),
    "offer": np.where(spam == 1, rng.poisson(2.4, size=n), rng.poisson(0.2, size=n)),
    "prediction": spam,
})
display(df.describe().round(2))
print("Random 5-row sample:")
display(df.sample(5, random_state=42))
print("Targeted query (free >= 3):")
display(df.query("free >= 3")["prediction"].value_counts(normalize=True))
```
