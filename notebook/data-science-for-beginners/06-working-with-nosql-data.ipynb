```python
!pip install -q tinydb duckdb pandas
import json
import duckdb
import pandas as pd
from tinydb import TinyDB, Query
from tinydb.storages import MemoryStorage
```

```python
# 1. TinyDB Document Store for PersonsData
db = TinyDB(storage=MemoryStorage)
persons_table = db.table("persons")
persons_table.insert_multiple([
    {"firstname": "Eva", "age": 44, "id": "8c74a315"},
    {"firstname": "Liam", "age": 28, "id": "9d12b401"},
    {"firstname": "Maya", "age": 34, "id": "7a88c112"},
    {"firstname": "Noah", "age": 51, "id": "3f44d908"},
])
Person = Query()
print("Persons younger than 40:", persons_table.search(Person.age < 40))

# 2. DuckDB SQL-over-JSON for Social Posts
posts = [
    {"id": "143278", "created_at": "2021-08-31T19:03:01Z", "text": "Exploring open-source data science in Python!"},
    {"id": "143279", "created_at": "2021-09-01T10:15:00Z", "text": "Building reproducible notebooks on Colab."},
    {"id": "143280", "created_at": "2021-09-02T14:22:00Z", "text": "Data cleaning with Pandas and DuckDB."}
]
with open("posts.json", "w") as f:
    json.dump(posts, f)

print("\nDuckDB JSON LIKE Query:")
display(duckdb.sql("SELECT id, text FROM read_json_auto('posts.json') WHERE text ILIKE '%data%'").df())
```
