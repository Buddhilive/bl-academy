# Lesson 08: Data Preparation, Imputation & Deduplication

In this notebook, you will:
1. Inspect the Iris dataset with `.info()`, `.head()`, and `.tail()`.
2. Distinguish `NaN` vs `None` vs `0` and `''` using `.isnull()`.
3. Practice `dropna(thresh=...)`, `ffill()`, `bfill()`, and `drop_duplicates()`.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
iris_df = pd.DataFrame(data=iris["data"], columns=iris["feature_names"])
iris_df.info()
display(iris_df.head())
```

```python
# Missing values: dropna, ffill, bfill
example2 = pd.DataFrame([[1, np.nan, 7], [2, 5, 8], [np.nan, 6, 9]])
example2[3] = np.nan
print("Rows with at least 3 non-null values:")
display(example2.dropna(axis="rows", thresh=3))

print("Forward-filled along columns:")
display(example2.ffill(axis=1))

# Deduplication
example4 = pd.DataFrame({"letters": ["A", "B"] * 2 + ["B"], "numbers": [1, 2, 1, 3, 3]})
print("After drop_duplicates(subset=['letters']):")
display(example4.drop_duplicates(["letters"]))
```
