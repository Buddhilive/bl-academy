# Lesson 05: Working with Relational Databases & SQLite3

In this notebook, you will:
1. Build normalized `cities` and `rainfall` tables with Primary and Foreign Keys in `sqlite3`.
2. Execute `SELECT`, `WHERE`, and `INNER JOIN` queries and load results into Pandas.
3. Query a self-contained `airports` relational schema.

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")
conn.executescript("""
CREATE TABLE cities (city_id INTEGER PRIMARY KEY, city TEXT, country TEXT);
CREATE TABLE rainfall (rainfall_id INTEGER PRIMARY KEY, city_id INTEGER, year INTEGER, amount INTEGER);
INSERT INTO cities VALUES (1, 'Tokyo', 'Japan'), (2, 'Atlanta', 'United States'), (3, 'Auckland', 'New Zealand');
INSERT INTO rainfall VALUES
    (1, 1, 2020, 1690), (2, 1, 2019, 1874), (3, 1, 2018, 1445),
    (4, 2, 2018, 1779), (5, 2, 2019, 1111), (6, 2, 2020, 1683),
    (7, 3, 2018, 1386), (8, 3, 2019, 942),  (9, 3, 2020, 1176);

CREATE TABLE airports (id INTEGER PRIMARY KEY, name TEXT, city TEXT, country TEXT, code TEXT);
INSERT INTO airports VALUES
    (1, 'Heathrow', 'London', 'United Kingdom', 'LHR'),
    (2, 'Gatwick', 'London', 'United Kingdom', 'LGW'),
    (3, 'Manchester', 'Manchester', 'United Kingdom', 'MAN'),
    (4, 'Haneda', 'Tokyo', 'Japan', 'HND'),
    (5, 'Hartsfield-Jackson', 'Atlanta', 'United States', 'ATL');
""")

df_rain = pd.read_sql_query("""
    SELECT cities.city, cities.country, rainfall.year, rainfall.amount
    FROM cities
    INNER JOIN rainfall ON cities.city_id = rainfall.city_id
    WHERE rainfall.year = 2019
""", conn)
display(df_rain)
```

```python
# Assignment: Airport queries
print("All UK Airports:")
display(pd.read_sql_query("SELECT name, city, code FROM airports WHERE country = 'United Kingdom'", conn))

print("London Airports:")
display(pd.read_sql_query("SELECT name, code FROM airports WHERE city = 'London'", conn))
```
