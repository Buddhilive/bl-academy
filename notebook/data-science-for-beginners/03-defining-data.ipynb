```python
import json
import pandas as pd

# Structured table
structured_df = pd.DataFrame({
    "customer_id": [1001, 1002, 1003],
    "phone_number": ["555-0101", "555-0102", "555-0103"],
    "quarterly_spend_usd": [420.50, 189.00, 912.75],
    "membership_tier": ["Gold", "Silver", "Platinum"]
})
print(structured_df.dtypes)

# Semi-structured JSON records with varying optional keys
semi_structured = [
    {"event": "watch_hr", "bpm": 72, "timestamp": "2026-11-21T08:00:00Z"},
    {"event": "watch_hr", "bpm": 135, "timestamp": "2026-11-21T08:15:00Z", "workout": "run"}
]
print("\nNormalized Semi-Structured JSON:")
print(pd.json_normalize(semi_structured))
```
