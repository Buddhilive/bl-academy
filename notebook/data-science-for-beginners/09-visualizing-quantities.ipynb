# Lesson 09: Visualizing Quantities with Matplotlib (Minnesota Birds)

In this notebook, you will:
1. Load a self-contained Minnesota birds dataset.
2. Spot and annotate the `MaxWingspan > 500` outliers (`Bald eagle` and `Prairie falcon`).
3. Plot horizontal category count bars and superimposed `MinLength`/`MaxLength` bars.

```python
import pandas as pd
import matplotlib.pyplot as plt

birds_data = [
    {"Name": "Black-bellied whistling-duck", "Category": "Ducks/Geese/Waterfowl", "MinLength": 47, "MaxLength": 56, "MaxWingspan": 94},
    {"Name": "Snow goose", "Category": "Ducks/Geese/Waterfowl", "MinLength": 64, "MaxLength": 79, "MaxWingspan": 165},
    {"Name": "Ross's goose", "Category": "Ducks/Geese/Waterfowl", "MinLength": 57.3, "MaxLength": 64, "MaxWingspan": 116},
    {"Name": "Greater white-fronted goose", "Category": "Ducks/Geese/Waterfowl", "MinLength": 64, "MaxLength": 81, "MaxWingspan": 165},
    {"Name": "Bald eagle", "Category": "Hawks/Eagles", "MinLength": 70, "MaxLength": 102, "MaxWingspan": 2300},
    {"Name": "Prairie falcon", "Category": "Falcons", "MinLength": 37, "MaxLength": 47, "MaxWingspan": 2500},
    {"Name": "Red-tailed hawk", "Category": "Hawks/Eagles", "MinLength": 45, "MaxLength": 65, "MaxWingspan": 133},
    {"Name": "Ruby-throated hummingbird", "Category": "Hummingbirds", "MinLength": 7, "MaxLength": 9, "MaxWingspan": 11},
    {"Name": "American white pelican", "Category": "Pelicans", "MinLength": 127, "MaxLength": 165, "MaxWingspan": 290},
]
birds = pd.DataFrame(birds_data)
display(birds.head())
```

```python
# 1. Annotate wingspan outliers (> 500 cm)
plt.figure(figsize=(8, 4))
plt.title("Max Wingspan in Centimeters (Outliers Highlighted)")
plt.ylabel("Wingspan (cm)")
plt.tick_params(axis="x", which="both", labelbottom=False, bottom=False)
for i, row in birds.iterrows():
    plt.plot(i, row["MaxWingspan"], "bo")
    if row["MaxWingspan"] > 500:
        plt.text(i + 0.1, row["MaxWingspan"] * 0.95, row["Name"], fontsize=10, color="red")
plt.show()

# 2. Filter outliers and plot superimposed Min/Max length by Category
clean = birds[~birds["Name"].isin(["Bald eagle", "Prairie falcon"])]
plt.figure(figsize=(8, 4))
plt.barh(clean["Category"], clean["MaxLength"], color="#A7452A", label="MaxLength")
plt.barh(clean["Category"], clean["MinLength"], color="#E07A5F", label="MinLength")
plt.xlabel("Length (cm)")
plt.title("Superimposed Min vs Max Bird Length by Category")
plt.legend()
plt.show()
```
