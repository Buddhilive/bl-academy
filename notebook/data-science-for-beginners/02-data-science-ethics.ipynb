```python
!pip install -q pandas scikit-learn fairlearn
import pandas as pd
import numpy as np
from fairlearn.metrics import MetricFrame, selection_rate, false_negative_rate
from sklearn.metrics import accuracy_score
```

```python
# 1. k-Anonymity Privacy Audit
df_patients = pd.DataFrame([
    {"zip": "94107", "age_band": "25-30", "condition": "Flu"},
    {"zip": "94107", "age_band": "25-30", "condition": "Cold"},
    {"zip": "10001", "age_band": "40-45", "condition": "Diabetes"},
    {"zip": "10001", "age_band": "40-45", "condition": "Hypertension"},
    {"zip": "60601", "age_band": "50-55", "condition": "Asthma"},
])
k_counts = df_patients.groupby(["zip", "age_band"]).size().rename("k_count")
print(k_counts)
print("\nRecords violating k >= 2 anonymity:")
print(k_counts[k_counts < 2])
```

```python
# 2. Fairlearn Subgroup Disparity Audit
y_true = np.array([1, 1, 1, 1, 0, 0, 1, 1, 1, 1, 0, 0])
y_pred = np.array([1, 1, 1, 0, 0, 0, 1, 0, 0, 0, 0, 0])
sensitive_group = np.array(["Group_A"] * 6 + ["Group_B"] * 6)

mf = MetricFrame(
    metrics={"accuracy": accuracy_score, "selection_rate": selection_rate, "fnr": false_negative_rate},
    y_true=y_true,
    y_pred=y_pred,
    sensitive_features=sensitive_group,
)
print("=== Fairness Metrics by Subgroup ===")
print(mf.by_group)
```
