In this notebook, you will:

2. Model Hot vs Cold storage tier costs across growing data volumes.

```python
import pandas as pd
import matplotlib.pyplot as plt

hot_ratios = [1.0, 0.5, 0.2, 0.1]
costs = [(10_000 * r * 0.023) + (10_000 * (1 - r) * 0.004) for r in hot_ratios]
df_cost = pd.DataFrame({"Hot_Share": [f"{int(r*100)}%" for r in hot_ratios], "Monthly_USD": costs})
display(df_cost)
df_cost.plot(x="Hot_Share", y="Monthly_USD", kind="bar", color="#A7452A", title="10 TB Monthly Storage Cost by Hot Tier Share")
plt.show()
```
