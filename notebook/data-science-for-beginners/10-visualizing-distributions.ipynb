# Lesson 10: Visualizing Distributions (Histograms & Seaborn KDE)

In this notebook, you will:
1. Build 1D and 2D histograms of bird body mass and length.
2. Overlay histograms by IUCN Conservation Status (`LC`, `NT`, `VU`, `EN`).
3. Plot smooth density curves and 2D contours with Seaborn `sns.kdeplot()`.

```python
!pip install -q numpy pandas matplotlib seaborn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(10)
n = 120
birds = pd.DataFrame({
    "MaxBodyMass": np.concatenate([rng.uniform(4, 58, 90), rng.uniform(120, 1800, 30)]),
    "MinLength": rng.uniform(10, 45, n),
    "MaxLength": rng.uniform(15, 65, n),
    "MinWingspan": rng.uniform(12, 95, n),
    "ConservationStatus": rng.choice(["LC", "NT", "VU", "EN"], size=n, p=[0.6, 0.2, 0.12, 0.08]),
    "Order": rng.choice(["Passeriformes", "Charadriiformes", "Anseriformes"], size=n),
})
filtered = birds[(birds["MaxBodyMass"] > 1) & (birds["MaxBodyMass"] < 60)]
display(filtered.head())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
filtered["MaxBodyMass"].plot(kind="hist", bins=20, ax=axes[0], color="#A7452A", edgecolor="white", title="Filtered MaxBodyMass (<60g)")
sns.kdeplot(data=filtered, x="MaxBodyMass", hue="Order", fill=True, common_norm=False, alpha=0.4, ax=axes[1])
axes[1].set_title("Seaborn KDE by Bird Order")
plt.tight_layout()
plt.show()

# 2D Histogram
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist2d(filtered["MaxBodyMass"], filtered["MaxLength"], bins=15, cmap="YlOrBr")
ax.set_xlabel("MaxBodyMass (g)")
ax.set_ylabel("MaxLength (cm)")
ax.set_title("2D Histogram: Mass vs Length")
plt.show()
```
