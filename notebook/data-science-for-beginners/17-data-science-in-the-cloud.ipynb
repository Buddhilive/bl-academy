# Lesson 17: Data Science in the Cloud with Google Colab & Gemini API

In this notebook, you will:
1. Analyze streaming social sentiment windows in **DuckDB**.
2. Extract structured clinical entities from scientific abstracts using **Google AI Studio Free Tier** (`google-genai` + `GEMINI_API_KEY`).

```python
!pip install -q duckdb pandas google-genai
import duckdb
import pandas as pd

events = pd.DataFrame([
    {"hashtag": "#datascience", "sentiment": 0.92, "ts": "2026-11-21 10:00:01"},
    {"hashtag": "#datascience", "sentiment": 0.78, "ts": "2026-11-21 10:00:15"},
    {"hashtag": "#python", "sentiment": 0.65, "ts": "2026-11-21 10:00:30"},
])
display(duckdb.sql("SELECT hashtag, COUNT(*) AS volume, ROUND(AVG(sentiment), 2) AS avg_sentiment FROM events GROUP BY hashtag").df())
```

```python
try:
    from google.colab import userdata
    from google import genai
    api_key = userdata.get("GEMINI_API_KEY")
    if api_key:
        client = genai.Client(api_key=api_key)
        resp = client.models.generate_content(
            model="gemini-2.5-flash",
            contents="Return JSON with keys 'treatments' and 'outcome' for: Patients treated with dexamethasone and supplemental oxygen showed reduced 28-day mortality."
        )
        print(resp.text)
    else:
        print("Add GEMINI_API_KEY to Colab Secrets to run live Gemini cloud extraction.")
except Exception as e:
    print(f"Gemini API check skipped: {e}")
```
