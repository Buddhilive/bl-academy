# Lesson 16: Data Communication & The 5 C's of Storytelling

In this notebook, you will:
1. Visualize the Emerson Mobile App weekend complaint surge (+42%) and 48-hour SLA impact.
2. Compare audience-tailored summary charts (Executive KPI vs Managerial SLA breakdown).

```python
import pandas as pd
import matplotlib.pyplot as plt

df_sla = pd.DataFrame({
    "Scenario": ["Current Baseline", "Weekend Reps + SLA Queue"],
    "Breach_Rate_Pct": [47.9, 5.6],
    "App_Store_Rating": [2.5, 4.0]
})
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
df_sla.plot(x="Scenario", y="Breach_Rate_Pct", kind="bar", ax=axes[0], color=["#57534E", "#A7452A"], legend=False, title="48h+ SLA Breach Rate (%)")
df_sla.plot(x="Scenario", y="App_Store_Rating", kind="bar", ax=axes[1], color=["#57534E", "#2E7D32"], legend=False, title="Projected App Store Rating")
plt.tight_layout()
plt.show()
```
