# Lesson 11: Visualizing Proportions (Pie, Donut & PyWaffle)

In this notebook, you will:
1. Group Audubon mushroom counts (`Edible` vs `Poisonous` and 7 habitats).
2. Build a **Pie Chart**, a **Donut Chart** (`plt.Circle`), and a **Waffle Chart** (`pywaffle`).

```python
!pip install -q pandas matplotlib pywaffle
import pandas as pd
import matplotlib.pyplot as plt
from pywaffle import Waffle

# Pie Chart: Edible vs Poisonous
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].pie([4208, 3916], labels=["Edible", "Poisonous"], autopct="%.1f %%", colors=["#2E7D32", "#A7452A"])
axes[0].set_title("Edible vs Poisonous Mushrooms")

# Donut Chart: Mushroom Habitats
habitats = ["Grasses", "Leaves", "Meadows", "Paths", "Urban", "Waste", "Wood"]
hab_counts = [2148, 832, 292, 1144, 368, 192, 3148]
axes[1].pie(hab_counts, labels=habitats, autopct="%1.1f%%", pctdistance=0.85)
axes[1].add_artist(plt.Circle((0, 0), 0.40, fc="white"))
axes[1].set_title("Mushroom Habitats (Donut)")
plt.tight_layout()
plt.show()
```

```python
# Waffle Chart of Mushroom Cap Colors
df_cap = pd.DataFrame({
    "color": ["brown", "buff", "cinnamon", "green", "pink", "purple", "red", "white", "yellow"],
    "amount": [2284, 168, 44, 16, 144, 16, 1500, 1040, 1072]
})
fig = plt.figure(
    FigureClass=Waffle,
    rows=15,
    values=df_cap["amount"],
    labels=list(df_cap["color"]),
    figsize=(12, 5),
    colors=["brown", "tan", "maroon", "green", "pink", "purple", "red", "lightgray", "gold"],
    legend={"loc": "upper left", "bbox_to_anchor": (1.02, 1)}
)
plt.title("Mushroom Cap Colors (Waffle Chart)")
plt.show()
```
