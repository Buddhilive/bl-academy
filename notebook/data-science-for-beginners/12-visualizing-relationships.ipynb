# Lesson 12: Visualizing Relationships (USDA Honey Production)

In this notebook, you will:
1. Explore USDA honey production (1998–2012) across states.
2. Build Seaborn `relplot` scatterplots (`hue` + `size`), `FacetGrid` small multiples, and `ax.twinx()` dual-axis line plots.

```python
!pip install -q pandas matplotlib seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

honey = pd.DataFrame([
    {"state": "AZ", "numcol": 55000, "yieldpercol": 60, "totalprod": 3300000, "priceperlb": 0.64, "year": 1998},
    {"state": "AZ", "numcol": 35000, "yieldpercol": 72, "totalprod": 2520000, "priceperlb": 1.34, "year": 2003},
    {"state": "AZ", "numcol": 25000, "yieldpercol": 64, "totalprod": 1600000, "priceperlb": 1.26, "year": 2008},
    {"state": "AZ", "numcol": 22000, "yieldpercol": 46, "totalprod": 1012000, "priceperlb": 1.79, "year": 2012},
    {"state": "CA", "numcol": 450000, "yieldpercol": 83, "totalprod": 37350000, "priceperlb": 0.62, "year": 1998},
    {"state": "CA", "numcol": 480000, "yieldpercol": 67, "totalprod": 32160000, "priceperlb": 1.39, "year": 2003},
    {"state": "CA", "numcol": 360000, "yieldpercol": 51, "totalprod": 18360000, "priceperlb": 1.39, "year": 2008},
    {"state": "CA", "numcol": 330000, "yieldpercol": 35, "totalprod": 11550000, "priceperlb": 1.94, "year": 2012},
])

sns.relplot(x="priceperlb", y="state", hue="year", size="year", palette="YlOrBr", data=honey, height=4, aspect=1.5)
plt.title("Honey Price per Pound by State & Year")
plt.show()
```

```python
# Dual-line plot with ax.twinx() and sns.despine()
fig, ax = plt.subplots(figsize=(9, 4))
sns.lineplot(x="year", y="numcol", data=honey, ax=ax, label="Number of Colonies", legend=False)
sns.despine()
ax.set_ylabel("# Colonies")
ax.set_title("Honey Colonies vs Yield per Colony")

ax2 = ax.twinx()
sns.lineplot(x="year", y="yieldpercol", data=honey, ax=ax2, color="#A7452A", label="Yield per Colony", legend=False)
sns.despine(right=False)
ax2.set_ylabel("Yield per Colony (lbs)")
fig.legend(loc="upper right", bbox_to_anchor=(0.88, 0.88))
plt.show()
```
