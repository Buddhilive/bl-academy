## Act 1 & Act 2: Research Agent + Scriptwriter Agent + HITL Approval Gate
Generate a structured two-host podcast script (`Speaker 1`, `Speaker 2`) grounded in local research facts.

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

def research_tool(topic: str) -> list[str]:
    return [
        f"{topic}: Small Language Models under 4B parameters run locally at 50-500ms latency.",
        "INT4 and GGUF Q4_K_M quantization cut memory footprint by ~75% with minimal quality loss."
    ]

facts = research_tool("Edge AI")
script_lines = [
    "Speaker 1: Welcome back to Future Bytes! Today we are exploring why Edge AI is moving off the cloud.",
    f"Speaker 2: Right! {facts[0]}",
    f"Speaker 1: And thanks to compression, {facts[1]}",
    "Speaker 2: That means fast, private AI right on your laptop or phone. Thanks for listening!"
]
podcast_script = "\n".join(script_lines)

# HITL Approval Gate (auto-approved in notebook; interactive in CLI)
approved = all(line.startswith(("Speaker 1:", "Speaker 2:")) for line in script_lines)
assert approved, "Script failed speaker formatting gate!"
with open("podcast.txt", "w") as f:
    f.write(podcast_script)
print("HITL Gate PASSED. Saved podcast.txt:\n" + podcast_script)
```

## Act 3: Multi-Speaker Podcast Audio Synthesis
Convert the approved `podcast.txt` script into a playable audio file (`future_bytes_episode1.mp3`) right inside Colab.

```python
from gtts import gTTS
from IPython.display import Audio, display

spoken_text = " ".join(line.split(": ", 1)[1] for line in script_lines)
tts = gTTS(text=spoken_text, lang="en")
tts.save("future_bytes_episode1.mp3")
print("Generated future_bytes_episode1.mp3!")
display(Audio("future_bytes_episode1.mp3"))
```
