```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

## 1. Quartiles, IQR Outliers & Correlation


```python
weights = np.array([180, 215, 210, 188, 176, 209, 200, 231, 185, 160, 197, 295], dtype=float)
heights = np.array([72, 75, 74, 71, 70, 74, 73, 77, 71, 68, 73, 79], dtype=float)

q1, med, q3 = np.percentile(weights, [25, 50, 75])
iqr = q3 - q1
upper_fence = q3 + 1.5 * iqr
r, p_val = stats.pearsonr(heights, weights)
print(f'Q1={q1:.1f}, Median={med:.1f}, Q3={q3:.1f}, IQR={iqr:.1f}, Upper Fence={upper_fence:.1f}')
print(f'Pearson r={r:.3f} (p={p_val:.4e})')
```

## 2. Diagnostic Histograms, Boxplots & Scatter Plots


```python
rng = np.random.default_rng(7)
study_hours = rng.uniform(1, 10, size=80)
exam_scores = 45 + 5.2 * study_hours + rng.normal(0, 4, size=80)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
axes[0].hist(exam_scores, bins=12, color='#A7452A', edgecolor='white')
axes[0].axvline(np.median(exam_scores), color='black', linestyle='--', label='Median')
axes[0].set_title('Exam Score Distribution')
axes[0].legend()

axes[1].boxplot(weights, vert=True)
axes[1].set_title('Weights Boxplot (IQR)')

axes[2].scatter(study_hours, exam_scores, color='#A7452A', alpha=0.8)
axes[2].set_title('Study Hours vs. Exam Score')
axes[2].set_xlabel('Hours')
axes[2].set_ylabel('Score')
plt.tight_layout()
plt.show()
```
