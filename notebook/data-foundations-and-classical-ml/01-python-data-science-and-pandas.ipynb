```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print('NumPy:', np.__version__, '| Pandas:', pd.__version__)
```

## 1. Index-Aligned Pandas Series


```python
dates = pd.date_range('2025-01-01', '2025-03-31', freq='D')
rng = np.random.default_rng(42)
items_sold = pd.Series(rng.integers(25, 50, size=len(dates)), index=dates)
weekly_promo = pd.Series(10, index=pd.date_range('2025-01-01', '2025-03-31', freq='W'))

# Safe index-aligned addition
total_items = items_sold.add(weekly_promo, fill_value=0)
print('First 5 days:\n', total_items.head())
```

## 2. DataFrame EDA & Monthly Resampling


```python
df = pd.DataFrame({
    'date': dates,
    'packs_sold': total_items.values,
    'region': rng.choice(['North', 'South', 'East'], size=len(dates)),
})

print(df.info())
print('\nSummary statistics:\n', df.describe())
print('\nBy region:\n', df.groupby('region')['packs_sold'].agg(['count', 'mean', 'max']))
```

```python
monthly = total_items.resample('ME').mean()
ax = monthly.plot(kind='bar', color='#A7452A', title='Monthly Average Packs Sold')
ax.set_ylabel('Average Packs')
plt.tight_layout()
plt.show()
```
