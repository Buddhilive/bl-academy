```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import root_mean_squared_error, r2_score, classification_report
```

## 1. Linear vs. Polynomial Regression on Seasonal Curve


```python
rng = np.random.default_rng(42)
month = np.linspace(1, 12, 120).reshape(-1, 1)
price = 25 - 4.0 * np.sin(month[:, 0] / 2.0) + rng.normal(0, 1.1, size=120)

X_tr, X_te, y_tr, y_te = train_test_split(month, price, test_size=0.25, random_state=42)
lin = LinearRegression().fit(X_tr, y_tr)
poly = make_pipeline(PolynomialFeatures(degree=3), LinearRegression()).fit(X_tr, y_tr)

for name, m in [('Linear', lin), ('Poly(deg=3)', poly)]:
    preds = m.predict(X_te)
    print(f'{name:12s} | RMSE={root_mean_squared_error(y_te, preds):.3f} | R2={r2_score(y_te, preds):.3f}')
```

## 2. Binary Classification with Logistic Regression


```python
from sklearn.datasets import make_classification
Xc, yc = make_classification(n_samples=250, n_features=4, random_state=42)
Xc_tr, Xc_te, yc_tr, yc_te = train_test_split(Xc, yc, test_size=0.25, stratify=yc, random_state=42)
clf = LogisticRegression().fit(Xc_tr, yc_tr)
print(classification_report(yc_te, clf.predict(Xc_te)))
```
