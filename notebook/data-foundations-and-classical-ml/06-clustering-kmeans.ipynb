```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

## 1. Elbow Method & Silhouette Score Sweep


```python
X, _ = make_blobs(n_samples=350, centers=4, cluster_std=0.85, random_state=42)
X_scaled = StandardScaler().fit_transform(X)

ks = list(range(2, 9))
inertias, silhouettes = [], []
for k in ks:
    km = KMeans(n_clusters=k, n_init='auto', random_state=42).fit(X_scaled)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_scaled, km.labels_))
    print(f'k={k} | inertia={km.inertia_:6.2f} | silhouette={silhouettes[-1]:.3f}')
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].plot(ks, inertias, marker='o', color='#A7452A')
ax[0].set_title('Elbow Method (Inertia)')
ax[0].set_xlabel('k')

ax[1].plot(ks, silhouettes, marker='o', color='#262626')
ax[1].set_title('Silhouette Score vs k')
ax[1].set_xlabel('k')
plt.tight_layout()
plt.show()
```
