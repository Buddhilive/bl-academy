In this notebook, you will build a lightweight Python **LLMOps DAG Tracer & 5-Metric Evaluation Harness** measuring:
- **Quality**, **Harm**, **Honesty (Groundedness)**, **Token Cost**, and **Latency (ms)** using `google-genai`.

```python
import os, time
from google import genai

def run_llmops_eval(query, context):
    t0 = time.perf_counter()
    key = os.environ.get("GEMINI_API_KEY")
    if key:
        client = genai.Client(api_key=key)
        resp = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=f"Context: {context}\nQuestion: {query}\nAnswer strictly from context:",
        )
        answer = resp.text.strip()
        in_tok = resp.usage_metadata.prompt_token_count
        out_tok = resp.usage_metadata.candidates_token_count
    else:
        answer = "LLMOps tracks Quality, Harm, Honesty, Cost, and Latency across DAG stages."
        in_tok, out_tok = 42, 16
    latency_ms = (time.perf_counter() - t0) * 1000
    cost = (in_tok * 0.15 + out_tok * 0.40) / 1_000_000
    return {"answer": answer, "latency_ms": round(latency_ms, 2), "cost_usd": round(cost, 6)}

print(run_llmops_eval("What does LLMOps track?", "LLMOps tracks Quality, Harm, Honesty, Cost, and Latency."))
```
