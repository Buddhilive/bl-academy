```python
import base64, os
from PIL import Image, ImageDraw
from google import genai

# Create self-contained sample.jpg
img = Image.new("RGB", (320, 200), (245, 240, 230))
draw = ImageDraw.Draw(img)
draw.rectangle([40, 80, 140, 170], fill=(167, 69, 42))
draw.ellipse([190, 60, 270, 140], fill=(30, 120, 180))
img.save("sample.jpg")

key = os.environ.get("GEMINI_API_KEY")
if key:
    client = genai.Client(api_key=key)
    resp = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=["Describe the shapes and colors in this image.", img],
    )
    print("Multimodal Analysis:", resp.text)
else:
    print("[Offline Demo Mode] Image contains a terracotta rectangle on the left and a blue circle on the right.")
```
