```python
import os
from google import genai

def get_client():
    key = None
    try:
        from google.colab import userdata
        key = userdata.get("GEMINI_API_KEY")
    except Exception:
        pass
    key = key or os.environ.get("GEMINI_API_KEY")
    return genai.Client(api_key=key) if key else None

client = get_client()
if client:
    emb = client.models.embed_content(
        model="gemini-embedding-001",
        contents="Compare encoder and decoder Transformer architectures.",
    )
    vec = emb.embeddings[0].values
    print(f"Embedding dimensions: {len(vec)} | First 5 values: {vec[:5]}")
else:
    print("[Offline Demo Mode] gemini-embedding-001 produces dense semantic vectors for RAG and search.")
```

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
print(classifier("Comparing open-weight and hosted LLMs makes architecture choices crystal clear!"))
```
