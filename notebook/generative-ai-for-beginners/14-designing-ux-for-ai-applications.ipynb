In this notebook, you will implement:
1. **Persona-Aware Socratic Scaffolding** (Student hints vs. Teacher rubrics).
2. **Graceful Out-of-Scope Handling** (History & Math tutor declining Geography).
3. **User Controls** (tone, length, and data opt-in state) with `google-genai`.

```python
import os
from google import genai
from google.genai import types

def tutor_respond(persona, subject, question, tone="encouraging"):
    if subject.lower() not in {"history", "math"}:
        return f"I'm currently trained to help with History and Math, so I can't reliably answer {subject} questions."
    system = (
        f"You are an AI tutor. Persona={persona}. Tone={tone}. "
        + ("Guide the student with a Socratic hint; do not give the final answer." if persona == "student" else "Provide a concise grading rubric.")
    )
    key = os.environ.get("GEMINI_API_KEY")
    if key:
        client = genai.Client(api_key=key)
        return client.models.generate_content(
            model="gemini-2.5-flash",
            contents=question,
            config=types.GenerateContentConfig(system_instruction=system),
        ).text
    return f"[Offline Demo Mode | {persona}] Hint: Break '{question}' into smaller steps first."

print(tutor_respond("student", "Math", "How do I solve 2x + 3 = 11?"))
print(tutor_respond("student", "Geography", "What is the capital of Peru?"))
```
