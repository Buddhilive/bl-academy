```python
import os
from google import genai
from google.genai import types

pitch = "We are building an AI coding tutor that runs Python in the browser and pairs with Colab notebooks."
personas = {
    "Pedagogy Lead": "Evaluate learning outcomes and ask 1 sharp follow-up question.",
    "Systems Architect": "Evaluate latency, browser WASM limits, and ask 1 technical question.",
    "Finance Officer": "Evaluate free-tier API sustainability and ask 1 unit-economics question.",
}

key = os.environ.get("GEMINI_API_KEY")
if key:
    client = genai.Client(api_key=key)
    for role, instr in personas.items():
        resp = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=f"Pitch: {pitch}",
            config=types.GenerateContentConfig(system_instruction=f"You are the {role}. {instr}", max_output_tokens=120),
        )
        print(f"=== {role} ===\n{resp.text.strip()}\n")
else:
    for role in personas:
        print(f"[Offline Demo Mode | {role}] Strong concept—how will you measure retention after Lesson 05?")
```
