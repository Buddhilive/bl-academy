```python
import os
from google import genai
from google.genai import types

def get_client():
    key = None
    try:
        from google.colab import userdata
        key = userdata.get("GEMINI_API_KEY")
    except Exception:
        pass
    key = key or os.environ.get("GEMINI_API_KEY")
    return genai.Client(api_key=key) if key else None

client = get_client()
```

```python
text = "Transformers use self-attention to weigh relationships between all tokens in a context window."

if client:
    count = client.models.count_tokens(model="gemini-2.5-flash", contents=text)
    print("Token count:", count.total_tokens)
    for temp in [0.0, 1.2]:
        resp = client.models.generate_content(
            model="gemini-2.5-flash",
            contents="Invent a 1-sentence tagline for an AI study tutor app.",
            config=types.GenerateContentConfig(temperature=temp, max_output_tokens=60),
        )
        print(f"T={temp}: {resp.text.strip()}")
else:
    print("[Offline Demo Mode] Estimated tokens:", len(text.split()))
    print("T=0.0: Your 24/7 AI tutor for clear, step-by-step learning.")
    print("T=1.2: Spark curiosity daily with an imaginative AI study companion!")
```
