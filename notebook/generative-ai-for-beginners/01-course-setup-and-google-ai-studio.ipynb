```python
import os
from google import genai

def get_gemini_client():
    api_key = None
    try:
        from google.colab import userdata
        api_key = userdata.get("GEMINI_API_KEY")
    except Exception:
        pass
    if not api_key:
        try:
            from dotenv import load_dotenv
            load_dotenv()
        except Exception:
            pass
        api_key = os.environ.get("GEMINI_API_KEY")
    if not api_key:
        print("GEMINI_API_KEY not found. Add it to Colab Secrets or .env to run live API calls.")
        return None
    return genai.Client(api_key=api_key)

client = get_gemini_client()
print("Client ready:" , client is not None)
```

```python
prompt = "Give me 3 concise reasons why developers should learn Generative AI engineering."

if client:
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
    )
    print(response.text)
else:
    print("[Offline Demo Mode] 1. Rapid prototyping with APIs. 2. Grounded RAG over private docs. 3. Structured tool calling.")
```
