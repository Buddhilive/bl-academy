```python
import os

GEMINI_API_KEY = None
try:
    from google.colab import userdata
    GEMINI_API_KEY = userdata.get("GEMINI_API_KEY")
except Exception:
    GEMINI_API_KEY = os.environ.get("GEMINI_API_KEY")

client = None
if GEMINI_API_KEY:
    from google import genai
    client = genai.Client(api_key=GEMINI_API_KEY)
print("Gemini API ready:", client is not None)
```

```python
knowledge_items = [
    {"rule": "Use SQLAlchemy 2.0", "in_repo": True},
    {"rule": "Run pytest before commit", "in_repo": True},
    {"rule": "OAuth2 required on /v2/", "in_repo": False},
    {"rule": "Max retry = 3", "in_repo": False},
]

in_repo_count = sum(1 for item in knowledge_items if item["in_repo"])
g_vis = 1.0 - (in_repo_count / len(knowledge_items))
print(f"Knowledge Visibility Gap G_vis = {g_vis:.2f} ({in_repo_count}/{len(knowledge_items)} in repo)")
assert g_vis == 0.5
```
