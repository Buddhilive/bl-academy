# Lesson 09: Task Boundaries, Little's Law & WIP=1 Scope Control

In this notebook, you will:
1. Model a **Scope Surface DAG** with dependencies.
2. Enforce `WIP = 1` and block new task activation whenever $\text{VCR} < 1.0$.

```python
tasks = [
    {"id": "F01", "deps": [], "state": "passing"},
    {"id": "F02", "deps": ["F01"], "state": "not_started"},
    {"id": "F03", "deps": ["F02"], "state": "not_started"},
]

def pick_next_task(items):
    active = [t for t in items if t["state"] == "active"]
    if active:
        raise RuntimeError(f"WIP=1 violation: {active[0]['id']} is still active!")
    done = {t["id"] for t in items if t["state"] == "passing"}
    for t in items:
        if t["state"] == "not_started" and all(d in done for d in t["deps"]):
            return t["id"]
    return None

nxt = pick_next_task(tasks)
print("Next eligible WIP=1 task:", nxt)
assert nxt == "F02"
```
