```python
import ast

def check_no_eval(code: str) -> str | None:
    for node in ast.walk(ast.parse(code)):
        if isinstance(node, ast.Call) and getattr(node.func, "id", None) == "eval":
            return "ERROR: eval() detected.\nWHY: Arbitrary code execution risk.\nFIX: Use ast.literal_eval() or json.loads()."
    return None

msg = check_no_eval("x = eval(user_input)")
print(msg)
assert msg and "FIX:" in msg
```
