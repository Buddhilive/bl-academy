# Lesson 10: Feature Lists as Harness Primitives & Pass-State Gating

In this notebook, you will:
1. Build a `feature_list.json` schema validator enforcing the `(behavior, verification, state)` triple.
2. Run a Pass-State Gate that rejects agent self-promotion and tracks **Back-Pressure**.

```python
import json

feature_list = [
    {"id": "F01", "behavior": "Chunk docs", "verification": "pytest -k chunk", "state": "passing", "evidence": "commit 11a"},
    {"id": "F02", "behavior": "Extract metadata", "verification": "pytest -k meta", "state": "active", "evidence": ""},
]

for item in feature_list:
    assert all(k in item and item[k] for k in ("behavior", "verification", "state")), f"Incomplete triple: {item}"

back_pressure = sum(1 for f in feature_list if f["state"] != "passing")
print(f"Schema valid. Back-pressure = {back_pressure} remaining feature(s).")
assert back_pressure == 1
```
