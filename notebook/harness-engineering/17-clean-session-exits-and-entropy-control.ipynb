```python
import tempfile
from pathlib import Path

ws = Path(tempfile.mkdtemp())
(ws / "debug-temp.log").write_text("noise", encoding="utf-8")
(ws / "app.py").write_text("def ok(): return True\n", encoding="utf-8")

# Idempotent cleanup: safe if run 1x or 10x
for p in ws.glob("debug-*.log"):
    p.unlink(missing_ok=True)

remaining = [p.name for p in ws.iterdir()]
print("Clean workspace files:", remaining)
assert remaining == ["app.py"]
```
