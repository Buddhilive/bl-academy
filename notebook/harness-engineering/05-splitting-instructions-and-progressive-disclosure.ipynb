```python
MAX_BYTES = 24_000
DEFAULT_RULES_BUDGET_TOKENS = 20_000

rules = [
    {"path": "AGENTS.md", "bytes": 3_200, "tokens": 850},
    {"path": ".agents/rules/a11y.md", "bytes": 4_100, "tokens": 1_100},
    {"path": ".agents/rules/huge-legacy.md", "bytes": 25_500, "tokens": 6_800},
    {"path": ".agents/rules/extra-guide.md", "bytes": 18_000, "tokens": 19_000},
]

used_tokens = 0
for r in rules:
    truncated = r["bytes"] > MAX_BYTES
    if used_tokens + r["tokens"] <= DEFAULT_RULES_BUDGET_TOKENS:
        mode = "INLINE"
        used_tokens += r["tokens"]
    else:
        mode = "POINTER_DEMOTED"
    print(f"{r['path']}: mode={mode}, truncated={truncated}")

assert used_tokens <= DEFAULT_RULES_BUDGET_TOKENS
```
