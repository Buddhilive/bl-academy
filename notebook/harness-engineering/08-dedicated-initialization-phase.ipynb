```python
import json

def session_start_hook(stdin_payload: dict) -> dict:
    ws = stdin_payload.get("workspacePaths", ["."])[0]
    msg = f"Workspace {ws} initialized. Read PROGRESS.md and run pytest -q before editing."
    return {"injectSteps": [{"systemMessage": {"systemMessage": msg}}]}

sample_stdin = {"conversationId": "conv-101", "workspacePaths": ["/workspace/kb-app"], "hookEventName": "SessionStart"}
out = session_start_hook(sample_stdin)
print(json.dumps(out, indent=2))
assert "injectSteps" in out
```
