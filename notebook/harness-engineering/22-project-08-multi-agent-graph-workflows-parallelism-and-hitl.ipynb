```python
from concurrent.futures import ThreadPoolExecutor

def run_tests():
    return {"check": "pytest", "ok": True}

def run_arch_review():
    return {"check": "architecture", "ok": True}

with ThreadPoolExecutor(max_workers=2) as pool:
    results = list(pool.map(lambda fn: fn(), [run_tests, run_arch_review]))

all_passed = all(r["ok"] for r in results)
print("Parallel Fan-In Results:", results, "-> all_passed=", all_passed)
assert all_passed
```
