```python
audit = {
    "Instructions": {"score": 4, "antigravity": "AGENTS.md + GEMINI.md + .agents/rules/"},
    "Tools": {"score": 5, "antigravity": "mcp_config.json + .agents/skills/"},
    "Environment": {"score": 3, "antigravity": "Standard Sandbox + pyproject.toml + init.sh"},
    "State": {"score": 2, "antigravity": "PROGRESS.md + feature_list.json + /compact"},
    "Feedback": {"score": 5, "antigravity": "pytest + .agents/hooks.json (PostToolUse/Stop)"},
}

lowest = min(audit.items(), key=lambda kv: kv[1]["score"])
print(f"Lowest-scoring subsystem: {lowest[0]} ({lowest[1]['score']}/5) -> Upgrade via {lowest[1]['antigravity']}")
assert lowest[0] == "State"
```
