```python
import json

def handle_stop_event(stdin_payload: dict, e2e_passed: bool) -> dict:
    if stdin_payload.get("continuationActive"):
        return {"decision": "allow"}
    if not e2e_passed:
        return {
            "decision": "continue",
            "reason": "E2E reset flow failed (500). Check EMAIL_HOST env var and templates/reset.html.",
        }
    return {"decision": "allow"}

res = handle_stop_event({"terminationReason": "NO_TOOL_CALL", "continuationActive": False}, e2e_passed=False)
print(json.dumps(res, indent=2))
assert res["decision"] == "continue"
```
