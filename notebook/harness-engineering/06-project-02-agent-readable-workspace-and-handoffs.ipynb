```python
import sqlite3, tempfile
from pathlib import Path

db_path = Path(tempfile.mkdtemp()) / "kb.sqlite"
with sqlite3.connect(db_path) as conn:
    conn.execute("CREATE TABLE IF NOT EXISTS docs (id TEXT PRIMARY KEY, title TEXT, body TEXT)")
    conn.execute("INSERT OR REPLACE INTO docs VALUES ('d1', 'System of Record', 'Put rules in the repo.')")
    conn.commit()

# Simulate fresh process reopening the database
with sqlite3.connect(db_path) as conn2:
    rows = conn2.execute("SELECT id, title, body FROM docs").fetchall()
    print("Reloaded after restart:", rows)
    assert len(rows) == 1

handoff_md = """# Session Handoff
- **Status**: Import, Detail View, and SQLite Persistence passing
- **Verification**: `pytest tests/test_kb.py` (3/3 passed)
- **Next Action**: Add document chunking and metadata extraction
"""
print(handoff_md)
```
