```python
class MultiTurnKB:
    def __init__(self):
        self.history: list[dict[str, str]] = []

    def ask(self, query: str) -> str:
        if "that" in query.lower() and self.history:
            prev_q = self.history[-1]["q"]
            ans = f"Follow-up to '{prev_q}': Keep WIP=1 and verify E2E. [doc#1]"
        else:
            ans = "Harness = Instructions, Tools, Environment, State, Feedback. [doc#0]"
        self.history.append({"q": query, "a": ans})
        return ans

kb = MultiTurnKB()
a1 = kb.ask("What is a harness?")
a2 = kb.ask("How do I verify that?")
print("Turn 1:", a1)
print("Turn 2:", a2)
assert "Follow-up" in a2 and "[doc#1]" in a2
```
