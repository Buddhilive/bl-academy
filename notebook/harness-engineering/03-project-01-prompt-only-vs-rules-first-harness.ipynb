```python
import json, os, tempfile
from pathlib import Path

GEMINI_API_KEY = None
try:
    from google.colab import userdata
    GEMINI_API_KEY = userdata.get("GEMINI_API_KEY")
except Exception:
    GEMINI_API_KEY = os.environ.get("GEMINI_API_KEY")

base_tmp = Path(tempfile.mkdtemp(prefix="p01_harness_"))
improved = base_tmp / "p01-improved"
(improved / "data").mkdir(parents=True, exist_ok=True)
(improved / "docs").mkdir(parents=True, exist_ok=True)
(improved / "data" / "welcome.md").write_text("Harness Engineering = Instructions + Tools + Environment + State + Feedback", encoding="utf-8")

agents_md = """# AGENTS.md
## Verification
- pytest -q
## Hard Constraints
- Never hardcode secrets; read GEMINI_API_KEY from environment.
"""
(improved / "AGENTS.md").write_text(agents_md, encoding="utf-8")

features = [
    {"id": "F01", "behavior": "App health returns OK", "status": "not_started"},
    {"id": "F02", "behavior": "List docs in data/", "status": "not_started"},
    {"id": "F03", "behavior": "Answer question from docs", "status": "not_started"},
    {"id": "F04", "behavior": "Ensure data/ exists", "status": "not_started"},
]
(improved / "feature_list.json").write_text(json.dumps(features, indent=2), encoding="utf-8")
print("Scaffolded p01-improved at:", improved)
assert (improved / "AGENTS.md").exists() and (improved / "feature_list.json").exists()
```
