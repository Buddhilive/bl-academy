```python
results = []
best_bpb = 1.250
for step, candidate in [(1, 1.234), (2, 1.240), (3, 1.219)]:
    kept = candidate < best_bpb
    if kept:
        best_bpb = candidate
    results.append({"step": step, "val_bpb": candidate, "status": "keep" if kept else "revert"})

print("Ratchet log:", results)
assert [r["status"] for r in results] == ["keep", "revert", "keep"]
```
