```python
import asyncio
import math
from mcp.server.fastmcp import FastMCP

mcp = FastMCP("ScientificCalculator")

@mcp.tool()
def add(a: float, b: float) -> float:
    """Add two numbers."""
    return a + b

@mcp.tool()
def divide(a: float, b: float) -> float:
    """Divide a by b with zero-division protection."""
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b

@mcp.tool()
def sqrt(x: float) -> float:
    """Compute square root for non-negative numbers."""
    if x < 0:
        raise ValueError("Cannot compute square root of negative number.")
    return math.sqrt(x)

@mcp.resource("calc://constants/euler")
def euler_constant() -> str:
    """Return Euler's number e."""
    return str(math.e)

@mcp.prompt()
def math_tutor_prompt(problem: str) -> str:
    """Guide step-by-step math explanations."""
    return f"Solve this step by step using calculator tools: {problem}"

async def inspect_server():
    tools = await mcp.list_tools()
    for t in tools:
        print(f"Tool: {t.name} | Schema: {t.inputSchema}")

await inspect_server()
```
