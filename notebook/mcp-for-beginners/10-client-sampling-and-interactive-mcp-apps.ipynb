```python
import os
from IPython.display import HTML, display
from mcp.server.fastmcp import FastMCP

mcp = FastMCP("WeatherAppServer")
UI_URI = "ui://weather/dashboard.html"

@mcp.resource(UI_URI, mime_type="text/html;profile=mcp-app")
def weather_ui() -> str:
    return """
    <div style="font-family:sans-serif;padding:16px;border:1px solid #A7452A;border-radius:8px;max-width:360px;">
      <h3 style="margin:0 0 8px;color:#A7452A;">Seattle Live Weather</h3>
      <p id="status">14 C - Light Rain (Humidity 82%)</p>
      <button onclick="document.getElementById('status').innerText='Refreshed: 15 C - Clearing'">Refresh Forecast</button>
    </div>
    """

display(HTML(weather_ui()))
```
