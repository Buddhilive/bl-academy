# Lesson 07: Linear & Polynomial Regression Four Ways

Build Simple Linear, Polynomial (`PolynomialFeatures(2)`), One-Hot `Variety`, and All-Features Polynomial Pipelines in Scikit-Learn using `np.sqrt(mean_squared_error(...))`.

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

rng = np.random.default_rng(42)
n = 120
varieties = rng.choice(['PIE TYPE', 'MINIATURE', 'FAIRYTALE'], size=n)
cities = rng.choice(['BALTIMORE', 'CHICAGO', 'BOSTON'], size=n)
packages = rng.choice(['1 1/9 bushel', '1/2 bushel'], size=n)
months = rng.choice([9, 10, 11], size=n)

base = np.where(varieties == 'MINIATURE', 34.0, np.where(varieties == 'FAIRYTALE', 26.0, 15.0))
price = base + np.where(months == 10, 2.5, 0.0) + np.where(cities == 'BOSTON', 1.8, 0.0) + rng.normal(0, 1.1, n)

df = pd.DataFrame({'Variety': varieties, 'Month': months, 'City': cities, 'Package': packages, 'Price': price})
X = pd.get_dummies(df['Variety']).join(df['Month']).join(pd.get_dummies(df['City'])).join(pd.get_dummies(df['Package']))
y = df['Price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
pipeline = make_pipeline(PolynomialFeatures(2), LinearRegression())
pipeline.fit(X_train, y_train)
pred = pipeline.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, pred))
print(f"RMSE: {rmse:.3f} ({rmse / pred.mean() * 100:.2f}%)")
print(f"Model determination (R2): {pipeline.score(X_train, y_train):.3f}")
```
