# Lesson 18: Translation & Sentiment Analysis (TextBlob + Google AI Studio Free Tier)

Analyze *Pride and Prejudice* sentiment with `TextBlob`, then use **Google AI Studio Free Tier** (`google-genai` + `GEMINI_API_KEY`) for idiomatic translation and sarcasm/irony-aware sentiment.

```python
!pip install -q textblob google-genai
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
```

```python
from textblob import TextBlob

quote1 = "It is a truth universally acknowledged, that a single man in possession of a good fortune, must be in want of a wife."
quote2 = "Darcy, as well as Elizabeth, really loved them; and they were both ever sensible of the warmest gratitude."
irony = "Our distress, my dear Lizzy, is very great."

for label, q in [("Opening", quote1), ("Closing", quote2), ("Irony", irony)]:
    print(f"{label:8s} -> {TextBlob(q).sentiment}")
```

## Google AI Studio Free Tier (`GEMINI_API_KEY`)

Add your free key from [aistudio.google.com/apikey](https://aistudio.google.com/apikey) to Colab Secrets as `GEMINI_API_KEY`, then run the cell below for idiomatic translation and irony-aware sentiment.

```python
import os
from google import genai

try:
    from google.colab import userdata
    api_key = userdata.get("GEMINI_API_KEY")
except Exception:
    api_key = os.environ.get("GEMINI_API_KEY")

if api_key:
    client = genai.Client(api_key=api_key)
    resp = client.models.generate_content(
        model="gemini-2.5-flash",
        contents="Translate idiomatically to French and classify true sentiment (accounting for irony): 'Our distress, my dear Lizzy, is very great.'"
    )
    print(resp.text)
else:
    print("Set GEMINI_API_KEY in Colab Secrets to run Gemini translation & irony analysis.")
```
