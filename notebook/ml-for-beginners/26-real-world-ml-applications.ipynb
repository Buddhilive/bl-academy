# Lesson 26: Classical Machine Learning in the Real World

Implement three real-world classical ML workflows in Scikit-Learn: Finance K-Means credit-card outlier detection, Healthcare Random Forest toxicity classification, and Journalism fake-news multi-classifier voting.

```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 1. Finance: K-Means Fraud Outlier Distance Scoring
rng = np.random.default_rng(26)
normal = rng.normal([50, 12], [15, 5], size=(100, 2))
km = KMeans(n_clusters=2, random_state=0, n_init='auto').fit(normal)
suspect = np.array([[1250.0, 820.0]])
min_dist = np.min(np.linalg.norm(km.cluster_centers_ - suspect, axis=1))
print(f"Suspect transaction distance from nearest cluster: {min_dist:.1f}")

# 2. Healthcare: Clinical Trial Toxicity Random Forest
X_med = rng.normal(0, 1, size=(200, 6))
y_med = ((X_med[:, 0] + 0.8 * X_med[:, 1]) > 0.3).astype(int)
rf = RandomForestClassifier(n_estimators=100, random_state=0).fit(X_med[:150], y_med[:150])
print(f"Clinical Trial Toxicity RF Accuracy: {accuracy_score(y_med[150:], rf.predict(X_med[150:]))*100:.1f}%")
```
