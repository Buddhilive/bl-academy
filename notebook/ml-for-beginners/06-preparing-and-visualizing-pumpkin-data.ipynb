# Lesson 06: Preparing & Visualizing North American Pumpkin Data

Self-contained USDA `US-pumpkins.csv` cleaning, bushel price normalization (`1 1/9` and `1/2` bushel), Matplotlib grouped bar plots, and Seaborn `relplot`, `catplot`, and correlation `heatmap`.

```python
import io
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

CSV_DATA = '''City Name,Package,Variety,Date,Low Price,High Price,Color
BALTIMORE,1 1/9 bushel cartons,PIE TYPE,09/24/2016,15.0,15.0,ORANGE
BALTIMORE,1 1/9 bushel cartons,PIE TYPE,10/01/2016,18.0,18.0,ORANGE
BALTIMORE,1/2 bushel cartons,MINIATURE,09/24/2016,15.0,15.0,ORANGE
BALTIMORE,1/2 bushel cartons,MINIATURE,10/08/2016,18.0,20.0,WHITE
CHICAGO,1 1/9 bushel cartons,PIE TYPE,09/17/2016,17.0,18.0,ORANGE
CHICAGO,1 1/9 bushel cartons,PIE TYPE,10/15/2016,16.0,17.0,ORANGE
CHICAGO,24 inch bins,HOWDEN TYPE,10/01/2016,130.0,150.0,ORANGE
BOSTON,1 1/9 bushel crates,PIE TYPE,11/05/2016,14.0,15.0,ORANGE
BOSTON,1/2 bushel cartons,MINIATURE,08/27/2016,14.0,14.0,ORANGE
'''

pumpkins = pd.read_csv(io.StringIO(CSV_DATA))
pumpkins = pumpkins[pumpkins['Package'].str.contains('bushel', case=True, regex=True)].copy()
price = (pumpkins['Low Price'] + pumpkins['High Price']) / 2
month = pd.DatetimeIndex(pumpkins['Date']).month

new_pumpkins = pd.DataFrame({
    'Month': month,
    'Package': pumpkins['Package'],
    'Low Price': pumpkins['Low Price'],
    'High Price': pumpkins['High Price'],
    'Price': price,
})
new_pumpkins.loc[new_pumpkins['Package'].str.contains('1 1/9'), 'Price'] = price / (1 + 1/9)
new_pumpkins.loc[new_pumpkins['Package'].str.contains('1/2'), 'Price'] = price / 0.5
print(new_pumpkins)

sns.catplot(x='Month', y='Price', data=new_pumpkins, kind='bar')
plt.title('Normalized Pumpkin Price per Bushel by Month')
plt.show()

corr = new_pumpkins[['Month', 'Low Price', 'High Price', 'Price']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Numeric Feature Correlation Matrix')
plt.show()
```
