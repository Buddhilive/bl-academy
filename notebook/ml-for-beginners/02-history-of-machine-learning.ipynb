# Lesson 02: The History of Machine Learning

Explore the foundational algorithms that shaped machine learning: **Bayes' Theorem (1763)**, **Legendre's Least Squares (1805)**, **Markov Chains (1913)**, **Rosenblatt's Perceptron (1957)**, and **Nearest Neighbor (1967)**.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1805 Ordinary Least Squares on noisy linear observations
rng = np.random.default_rng(42)
x = np.linspace(0, 10, 25)
y = 2.0 + 1.5 * x + rng.normal(0, 1.0, size=len(x))

b, a = np.polyfit(x, y, 1)
print(f"Fitted OLS line: Y = {a:.2f} + {b:.2f}X")

plt.figure(figsize=(7, 4))
plt.scatter(x, y, label='Observations', color='#1F2937')
plt.plot(x, a + b * x, color='#A7452A', linewidth=2.5, label='1805 Least Squares Fit')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Legendre Method of Least Squares (1805)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 1967 Nearest Neighbor Classifier

Cover and Hart formalized Nearest Neighbor classification in 1967. Let's run it with Scikit-Learn's `KNeighborsClassifier`.

```python
from sklearn.neighbors import KNeighborsClassifier

X_train = np.array([[1, 1], [1.5, 2], [5, 5], [6, 5.5]])
y_train = np.array([0, 0, 1, 1])

knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_train, y_train)
print("Prediction for [5.2, 4.9]:", knn.predict([[5.2, 4.9]])[0])
```
