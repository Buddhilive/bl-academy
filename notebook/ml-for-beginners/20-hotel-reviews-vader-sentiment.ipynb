# Lesson 20: Hotel Reviews Tag Filtering, NLTK VADER & Gemini Sarcasm Audit

Normalize city addresses, encode 8 binary guest/trip tags, filter stop words, score sentiment with `NLTK VADER`, and compare sarcastic reviews using **Google AI Studio Free Tier** (`google-genai`).

```python
!pip install -q nltk pandas google-genai
import nltk
nltk.download('stopwords', quiet=True)
nltk.download('vader_lexicon', quiet=True)
```

```python
import pandas as pd
from nltk.corpus import stopwords
from nltk.sentiment.vader import SentimentIntensityAnalyzer

df = pd.DataFrame([
    {'Hotel_Name': 'Westminster Inn', 'Hotel_Address': 'London, United Kingdom', 'Tags': "[' Leisure trip ', ' Couple ']", 'Negative_Review': 'No Negative', 'Positive_Review': 'Amazing breakfast and lovely view'},
    {'Hotel_Name': 'Paris Loft', 'Hotel_Address': 'Paris, France', 'Tags': "[' Business trip ', ' Solo traveler ']", 'Negative_Review': 'Of course I LOVED sleeping in a room with no heating', 'Positive_Review': 'No Positive'},
])

df['Leisure_trip'] = df['Tags'].apply(lambda t: 1 if 'Leisure trip' in t else 0)
df['Business_trip'] = df['Tags'].apply(lambda t: 1 if 'Business trip' in t else 0)

cache = set(stopwords.words('english'))
vader = SentimentIntensityAnalyzer()

def calc_sentiment(text):
    if text in ('No Negative', 'No Positive'):
        return 0.0
    cleaned = ' '.join(w for w in text.split() if w.lower() not in cache)
    return vader.polarity_scores(cleaned)['compound']

df['Negative_Sentiment'] = df['Negative_Review'].apply(calc_sentiment)
df['Positive_Sentiment'] = df['Positive_Review'].apply(calc_sentiment)
print(df[['Hotel_Name', 'Negative_Sentiment', 'Positive_Sentiment']])
```
