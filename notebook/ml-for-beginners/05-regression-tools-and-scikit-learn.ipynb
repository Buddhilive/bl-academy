# Lesson 05: Getting Started with Python & Scikit-Learn for Regression

Load the built-in **Diabetes Dataset** (`442` samples, `10` features), slice `bmi` (`X[:, 2]`), fit a `LinearRegression` model, and plot the regression line.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model, model_selection

X, y = datasets.load_diabetes(return_X_y=True)
print("Original shape:", X.shape)
print("First sample:", X[0])

# Select 3rd feature (BMI) and reshape to 2D
X_bmi = X[:, 2].reshape((-1, 1))
X_train, X_test, y_train, y_test = model_selection.train_test_split(X_bmi, y, test_size=0.33, random_state=42)

model = linear_model.LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

plt.figure(figsize=(8, 5))
plt.scatter(X_test, y_test, color='black', label='Actual')
plt.plot(X_test, y_pred, color='#A7452A', linewidth=3, label='Linear Regression')
plt.xlabel('Scaled BMIs')
plt.ylabel('Disease Progression')
plt.title('Diabetes Progression Against BMI')
plt.legend()
plt.show()
```
