# Lesson 21: Introduction to Time Series Forecasting (World Power Usage)

Generate a self-contained GEFCom2014-style hourly electricity load dataset (`2012–2014`) and visualize full multi-year seasonality alongside the July 1–7, 2014 diurnal window.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(21)
idx = pd.date_range('2012-01-01', '2014-12-31 23:00:00', freq='h')
hours = np.arange(len(idx))
seasonal_year = 450 * np.cos(2 * np.pi * idx.dayofyear / 365.25)
diurnal = 550 * np.sin(2 * np.pi * (idx.hour - 7) / 24.0)
trend = 0.015 * hours
load = 3100 + trend + seasonal_year + diurnal + rng.normal(0, 75, len(idx))

energy = pd.DataFrame({'load': load}, index=idx)
energy.to_csv('energy.csv')
print(energy.head())

energy['2014-07-01':'2014-07-07'].plot(y='load', figsize=(12, 4), color='#A7452A')
plt.title('Hourly Electricity Load: First Week of July 2014')
plt.xlabel('timestamp')
plt.ylabel('load (MW)')
plt.show()
```
