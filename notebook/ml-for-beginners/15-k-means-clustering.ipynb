# Lesson 15: K-Means Clustering, Silhouette Score & Elbow Method

Run `KMeans` with `k-means++`, plot the WCSS Elbow curve for `k=1..10`, compute `silhouette_score`, and compare unscaled vs `StandardScaler` clustering.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

rng = np.random.default_rng(15)
X = np.vstack([
    rng.normal([15, 0.72, 0.25, -5.0, 0.75], [5, 0.08, 0.1, 1.0, 0.1], (80, 5)),
    rng.normal([35, 0.78, 0.20, -4.2, 0.82], [6, 0.07, 0.1, 0.9, 0.08], (80, 5)),
    rng.normal([58, 0.66, 0.35, -6.1, 0.68], [5, 0.09, 0.1, 1.1, 0.1], (80, 5)),
])

wcss = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init='auto').fit(X)
    wcss.append(km.inertia_)

plt.plot(range(1, 11), wcss, marker='o', color='#A7452A')
plt.title('Elbow Method (WCSS Inertia)')
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS')
plt.show()

km3 = KMeans(n_clusters=3, random_state=0, n_init='auto').fit(X)
print('Unscaled Silhouette Score:', round(silhouette_score(X, km3.labels_), 3))

X_scaled = StandardScaler().fit_transform(X)
km3_s = KMeans(n_clusters=3, random_state=0, n_init='auto').fit(X_scaled)
print('Scaled Silhouette Score:', round(silhouette_score(X_scaled, km3_s.labels_), 3))
```
