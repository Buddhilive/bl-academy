# Lesson 19: Exploring European Hotel Reviews (EDA & Vectorization)

Perform exploratory data analysis on European Hotel Reviews: inspect `Reviewer_Nationality`, compute `Calc_Average_Score`, and benchmark vectorized boolean sums against `.apply(lambda ...)`.

```python
import time
import numpy as np
import pandas as pd

rng = np.random.default_rng(19)
n = 20000
hotels = ['Britannia International Hotel Canary Wharf', 'Park Plaza Westminster', 'Hotel Da Vinci', 'Kube Hotel Ice Bar']
df = pd.DataFrame({
    'Hotel_Name': rng.choice(hotels, size=n),
    'Reviewer_Nationality': rng.choice(['United Kingdom', 'United States of America', 'Australia', 'France'], size=n, p=[0.5, 0.2, 0.15, 0.15]),
    'Average_Score': 7.5,
    'Reviewer_Score': np.round(rng.uniform(4.5, 9.8, size=n), 1),
    'Negative_Review': rng.choice(['No Negative', 'Small room', 'Noisy street'], size=n, p=[0.25, 0.4, 0.35]),
    'Positive_Review': rng.choice(['No Positive', 'Great breakfast', 'Helpful staff'], size=n, p=[0.1, 0.5, 0.4]),
})

df['Calc_Average_Score'] = df.groupby('Hotel_Name')['Reviewer_Score'].transform('mean').round(1)
print(df[['Hotel_Name', 'Average_Score', 'Calc_Average_Score']].drop_duplicates())

t0 = time.time()
v_count = int((df['Negative_Review'] == 'No Negative').sum())
print(f"Vectorized 'No Negative' count: {v_count} in {time.time() - t0:.4f}s")
```
