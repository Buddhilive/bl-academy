Run tokenization, N-grams, `ConllExtractor` noun-phrase chunking, inflection, lemmatization, and the upgraded Sympathetic Marvin bot using `TextBlob` and `NLTK`.

```python
!pip install -q textblob nltk
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('brown', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('conll2000', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
```

```python
from textblob import TextBlob, Word
from textblob.np_extractors import ConllExtractor

extractor = ConllExtractor()
sample = TextBlob("I went for a walk and saw a lovely cat", np_extractor=extractor)
print("Noun phrases:", sample.noun_phrases)
print("Polarity:", sample.sentiment.polarity)
print("N-grams (n=3):", sample.ngrams(n=3)[:3])
print("Lemma of 'flying':", Word("flying").lemmatize("v"))
```
