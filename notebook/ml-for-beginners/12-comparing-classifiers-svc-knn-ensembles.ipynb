# Lesson 12: Comparing Classifiers (Linear SVC, KNN, SVC, Random Forest & AdaBoost)

Walk the Scikit-Learn estimator map and benchmark five classifier families side by side.

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

rng = np.random.default_rng(12)
X = rng.integers(0, 2, size=(600, 15))
y = np.repeat(['chinese', 'indian', 'japanese', 'korean', 'thai'], 120)
for i in range(5):
    X[i*120:(i+1)*120, i*3:i*3+2] = 1

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

classifiers = {
    'Linear SVC': SVC(kernel='linear', C=10, probability=True, random_state=0),
    'KNN': KNeighborsClassifier(10),
    'RBF SVC': SVC(),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=0),
    'AdaBoost': AdaBoostClassifier(n_estimators=100, random_state=0),
}

for name, clf in classifiers.items():
    clf.fit(X_train, y_train)
    acc = accuracy_score(y_test, clf.predict(X_test))
    print(f"{name:15s}: {acc * 100:.1f}%")
```
