# Lesson 16: Introduction to NLP, ELIZA & Marvin the Bot

Implement both Joseph Weizenbaum's 1966 **ELIZA** pronoun-reflection rules and **Marvin the Simple Robot** in Python.

```python
import random
import re

random_responses = [
    "That is quite interesting, please tell me more.",
    "I see. Do go on.",
    "Why do you say that?",
    "Funny weather we've been having, isn't it?",
    "Let's change the subject.",
    "Did you catch the game last night?",
]

def marvin_chat(turns):
    print("Hello, I am Marvin, the simple robot.")
    for msg in turns:
        print(f"> {msg}")
        if msg.strip().lower() == "bye":
            print("It was nice talking to you, goodbye!")
            break
        print(random.choice(random_responses))

marvin_chat(["I am good thanks", "I like classical ML", "bye"])
```
