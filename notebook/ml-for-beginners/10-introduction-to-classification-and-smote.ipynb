# Lesson 10: Introduction to Classification & SMOTE Balancing

Explore the Asian and Indian Cuisines dataset, remove ubiquitous confounding ingredients (`rice`, `garlic`, `ginger`), and balance all 5 classes with `imblearn.over_sampling.SMOTE`.

```python
import numpy as np
import pandas as pd
from imblearn.over_sampling import SMOTE

rng = np.random.default_rng(7)
class_sizes = {'korean': 100, 'indian': 75, 'chinese': 55, 'japanese': 40, 'thai': 36}
ingredients = ['rice', 'garlic', 'ginger', 'kimchi', 'gochujang', 'garam_masala', 'turmeric', 'soy_sauce', 'mirin', 'lemongrass', 'fish_sauce']

rows = []
for cuisine, n in class_sizes.items():
    for _ in range(n):
        row = {ing: int(rng.random() < 0.08) for ing in ingredients}
        row['rice'] = int(rng.random() < 0.7)
        row['garlic'] = int(rng.random() < 0.65)
        row['ginger'] = int(rng.random() < 0.55)
        if cuisine == 'korean': row['kimchi'] = 1; row['gochujang'] = 1
        elif cuisine == 'indian': row['garam_masala'] = 1; row['turmeric'] = 1
        elif cuisine == 'chinese': row['soy_sauce'] = 1
        elif cuisine == 'japanese': row['mirin'] = 1; row['soy_sauce'] = 1
        elif cuisine == 'thai': row['lemongrass'] = 1; row['fish_sauce'] = 1
        row['cuisine'] = cuisine
        rows.append(row)

df = pd.DataFrame(rows)
print('Before SMOTE:\n', df['cuisine'].value_counts())

X = df.drop(['cuisine', 'rice', 'garlic', 'ginger'], axis=1)
y = df['cuisine']
X_res, y_res = SMOTE(random_state=42).fit_resample(X, y)
print('\nAfter SMOTE:\n', y_res.value_counts())
```
