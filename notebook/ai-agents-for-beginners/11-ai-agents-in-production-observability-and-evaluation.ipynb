```python
from opentelemetry import trace
from opentelemetry.sdk.trace import TracerProvider
from opentelemetry.sdk.trace.export import SimpleSpanProcessor, ConsoleSpanExporter

provider = TracerProvider()
provider.add_span_processor(SimpleSpanProcessor(ConsoleSpanExporter()))
trace.set_tracer_provider(provider)
tracer = trace.get_tracer("expense_claim_agent")

with tracer.start_as_current_span("expense_claim_run") as root:
    root.set_attribute("claim.id", "EXP-1042")
    with tracer.start_as_current_span("policy_check") as child:
        child.set_attribute("routed.model", "gemini-2.5-flash-lite")
        print("Claim EXP-1042 validated within meal limit.")
```
