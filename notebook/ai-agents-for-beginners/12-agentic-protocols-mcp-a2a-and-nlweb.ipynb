```python
from dataclasses import dataclass
import chromadb

@dataclass
class AgentCard:
    name: str
    skills: list[str]
    endpoint: str

hotel_card = AgentCard("HotelAgent", ["search_hotels", "book_hotel"], "https://hotel-agent.example/a2a")
print("Discovered A2A AgentCard:", hotel_card)

chroma = chromadb.Client()
nlweb_col = chroma.get_or_create_collection("nlweb_honolulu")
nlweb_col.upsert(
    ids=["h1"],
    documents=["OceanBreeze Resort in Honolulu: family-friendly hotel with pool and vegan restaurant."]
)
print("NLWeb ask():", nlweb_col.query(query_texts=["vegan restaurant hotel Honolulu"], n_results=1)["documents"][0][0])
```
