```python
import os
from google import genai
from google.genai import types

def get_gemini_client():
    api_key = None
    try:
        from google.colab import userdata
        api_key = userdata.get("GEMINI_API_KEY")
    except Exception:
        pass
    api_key = api_key or os.environ.get("GEMINI_API_KEY")
    return genai.Client(api_key=api_key) if api_key else None

client = get_gemini_client()
meta_prompt = "You are an expert at creating AI agent system prompts. Structure the output with Objective, Key Responsibilities, Tone, and Guardrails."
basic_prompt = "You are a travel agent for Contoso Travel that looks up flights, books seats, and alerts on delays."

if client:
    res = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=basic_prompt,
        config=types.GenerateContentConfig(system_instruction=meta_prompt)
    )
    print(res.text[:600])
else:
    print("[Offline Simulation] Generated Structured System Prompt: Objective + 5 Responsibilities + Tone + Guardrails.")
```
