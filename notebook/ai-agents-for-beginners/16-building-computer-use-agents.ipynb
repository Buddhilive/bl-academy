```python
from pydantic import BaseModel

class Listing(BaseModel):
    title: str
    nightly_price_usd: int
    rating: float

sample_listings = [
    Listing(title="Södermalm Loft", nightly_price_usd=92, rating=4.9),
    Listing(title="Gamla Stan Studio", nightly_price_usd=118, rating=4.8),
    Listing(title="Vasastan Flat", nightly_price_usd=105, rating=4.7),
]
cheapest = min(sample_listings, key=lambda x: x.nightly_price_usd)
print("Cheapest Stockholm stay:", cheapest.model_dump_json())
```
