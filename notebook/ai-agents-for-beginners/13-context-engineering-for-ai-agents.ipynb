```python
import chromadb

chroma = chromadb.Client()
tools_col = chroma.get_or_create_collection("tool_loadout")
tools_col.upsert(
    ids=["book_flight", "public_transport_info", "currency_converter"],
    documents=[
        "Book international airline flights between cities.",
        "Get metro, bus, and train transit directions inside a city like Paris.",
        "Convert currency exchange rates."
    ]
)
res = tools_col.query(query_texts=["What is the best way to get around Paris?"], n_results=1)
print("Selected Tool Loadout:", res["ids"][0])
```
