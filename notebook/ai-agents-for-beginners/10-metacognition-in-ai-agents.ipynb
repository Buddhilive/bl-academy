```python
import sqlite3

class HotelRecommendationAgent:
    def __init__(self):
        self.strategy = "cheapest"
        self.history = []

    def recommend(self, hotels):
        pick = min(hotels, key=lambda x: x["price"]) if self.strategy == "cheapest" else max(hotels, key=lambda x: x["quality"])
        self.history.append((self.strategy, pick))
        return pick

    def reflect(self):
        strat, last = self.history[-1]
        if last["price"] < 100 or last["quality"] < 7:
            self.strategy = "highest_quality"
            return f"Switched strategy from {strat} -> {self.strategy}"
        return "Strategy performing well."

hotels = [
    {"name": "Budget Inn", "price": 80, "quality": 6},
    {"name": "Comfort Suites", "price": 120, "quality": 8},
    {"name": "Luxury Stay", "price": 200, "quality": 9},
]
agent = HotelRecommendationAgent()
print("Initial:", agent.recommend(hotels))
print("Reflection:", agent.reflect())
print("Adjusted:", agent.recommend(hotels))
```
