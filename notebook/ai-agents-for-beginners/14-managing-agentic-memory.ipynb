1. Build a **Mem0-style Two-Phase Memory Pipeline** (`ADD`, `UPDATE`, `DELETE`) backed by `ChromaDB`.
2. Build a **Cognee-style Knowledge Graph** (`networkx` + vector embeddings) for hybrid entity retrieval.

```python
import networkx as nx
import chromadb

G = nx.DiGraph()
G.add_edge("Ben", "Mountain-View Coffee", relation="PREFERS")
G.add_edge("Ben", "Advanced Ski Slopes", relation="AVOIDS_DUE_TO_INJURY")

for u, v, data in G.edges(data=True):
    print(f"({u}) -[{data['relation']}]-> ({v})")

chroma = chromadb.Client()
mem_col = chroma.get_or_create_collection("ben_memory")
mem_col.upsert(ids=["m1"], documents=["Ben avoids advanced ski slopes due to a past injury."])
print("Vector Recall:", mem_col.query(query_texts=["Which ski slopes for Ben?"], n_results=1)["documents"][0][0])
```
