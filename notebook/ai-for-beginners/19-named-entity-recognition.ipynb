# Lesson 19: Named Entity Recognition (NER) with PyTorch & Transformers

In this notebook, you will:
1. Run a pre-trained **BERT Token Classification** NER pipeline using HuggingFace `transformers`.
2. Complete the **Lab**: Extract medical/news entities and inspect BIO spans.

```python
!pip install -q transformers
from transformers import pipeline

ner = pipeline("ner", model="dslim/bert-base-NER", aggregation_strategy="simple")
text = "Alan Turing worked at Bletchley Park in the United Kingdom and published his famous paper in 1950."
entities = ner(text)
for ent in entities:
    print(f"{ent['word']:20s} | {ent['entity_group']:5s} | score={ent['score']:.3f}")
```

```python
# Lab:Inspect raw BIO token predictions using AutoTokenizer & AutoModelForTokenClassification
import torch
from transformers import AutoTokenizer, AutoModelForTokenClassification

tok = AutoTokenizer.from_pretrained("dslim/bert-base-NER")
model = AutoModelForTokenClassification.from_pretrained("dslim/bert-base-NER")

inputs = tok("Hugging Face is based in New York City.", return_tensors="pt")
with torch.no_grad():
    logits = model(**inputs).logits
preds = logits.argmax(-1)[0].tolist()
for token, pred_id in zip(tok.convert_ids_to_tokens(inputs["input_ids"][0]), preds):
    print(f"{token:12s} -> {model.config.id2label[pred_id]}")
```
