# Lesson 18: Attention Mechanisms, Transformers & BERT

In this notebook, you will:
1. Implement **Scaled Dot-Product Self-Attention** in PyTorch.
2. Load a pre-trained **DistilBERT** model via HuggingFace `transformers` for sentiment classification.

```python
import math
import torch
import torch.nn.functional as F

def scaled_dot_product_attention(Q, K, V):
    d_k = Q.size(-1)
    scores = (Q @ K.transpose(-2, -1)) / math.sqrt(d_k)
    weights = F.softmax(scores, dim=-1)
    return weights @ V, weights

Q = K = V = torch.randn(2, 5, 16)  # (batch=2, seq_len=5, d_k=16)
out, attn = scaled_dot_product_attention(Q, K, V)
print("Attention output shape:", tuple(out.shape), "Weights shape:", tuple(attn.shape))
```
