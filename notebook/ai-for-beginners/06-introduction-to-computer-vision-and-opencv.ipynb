```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create a synthetic grayscale test image with a geometric pattern and noise
img = np.zeros((160, 240, 3), dtype=np.uint8)
cv2.circle(img, (80, 80), 30, (255, 180, 60), -1)
cv2.rectangle(img, (140, 40), (210, 120), (60, 220, 120), -1)

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 50, 150)

fig, ax = plt.subplots(1, 3, figsize=(12, 3))
ax[0].imshow(rgb); ax[0].set_title("RGB Image")
ax[1].imshow(gray, cmap="gray"); ax[1].set_title("Grayscale")
ax[2].imshow(edges, cmap="gray"); ax[2].set_title("Canny Edges")
for a in ax: a.axis("off")
plt.show()
```

## Lab: Video Frame Differencing & Optical Flow Motion Tracking

We generate a short synthetic video of a moving circle and track its displacement using both frame differencing and Lucas-Kanade optical flow.

```python
frames = []
for t in range(10):
    f = np.zeros((120, 160), dtype=np.uint8)
    cv2.circle(f, (30 + t * 10, 60), 12, 255, -1)
    frames.append(f)

diff = cv2.absdiff(frames[1], frames[0])
print(f"Frame 0->1 changed pixels: {(diff > 25).sum()}")

# Lucas-Kanade sparse optical flow
p0 = cv2.goodFeaturesToTrack(frames[0], maxCorners=10, qualityLevel=0.3, minDistance=7)
p1, st, err = cv2.calcOpticalFlowPyrLK(frames[0], frames[1], p0, None)
if p1 is not None:
    dx = (p1[st == 1][:, 0] - p0[st == 1][:, 0]).mean()
    print(f"Estimated horizontal motion dx: {dx:.2f} px/frame")
```
