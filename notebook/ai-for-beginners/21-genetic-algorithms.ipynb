In this notebook, you will:
1. Solve the **Diophantine Equation** ($a + 2b + 3c + 4d = 30$) with a Genetic Algorithm.
2. Solve the classic **8-Queens Puzzle** using permutation chromosomes, crossover, and mutation.

```python
import random
random.seed(7)

# 8-Queens: chromosome[col] = row (0..7), minimizing diagonal conflicts
def conflicts(board):
    n = len(board)
    clashes = 0
    for i in range(n):
        for j in range(i + 1, n):
            if abs(board[i] - board[j]) == abs(i - j):
                clashes += 1
    return clashes

pop = [random.sample(range(8), 8) for _ in range(40)]
for gen in range(500):
    pop.sort(key=conflicts)
    if conflicts(pop[0]) == 0:
        print(f"8-Queens solved in generation {gen}: {pop[0]}")
        break
    next_pop = pop[:8]
    while len(next_pop) < 40:
        p = list(random.choice(pop[:15]))
        i, j = random.sample(range(8), 2)
        p[i], p[j] = p[j], p[i]
        next_pop.append(p)
    pop = next_pop
```

```python
# Fair Treasure Division (Partition Problem) with Binary Chromosomes
treasures = [12, 5, 8, 21, 14, 9, 3, 18]

def split_diff(mask):
    a = sum(v for v, m in zip(treasures, mask) if m == 0)
    b = sum(v for v, m in zip(treasures, mask) if m == 1)
    return abs(a - b)

pop = [[random.randint(0, 1) for _ in treasures] for _ in range(30)]
for _ in range(100):
    pop.sort(key=split_diff)
    if split_diff(pop[0]) == 0:
        break
    pop = pop[:6] + [
        [g ^ (1 if random.random() < 0.2 else 0) for g in random.choice(pop[:6])]
        for _ in range(24)
    ]
print(f"Best treasure split difference: {split_diff(pop[0])} | Mask: {pop[0]}")
```
