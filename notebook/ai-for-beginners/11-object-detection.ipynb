# Lesson 11: Object Detection with Torchvision

In this notebook, you will:
1. Implement **IoU** and **Non-Maximum Suppression (NMS)** with `torchvision.ops.nms`.
2. Run a pre-trained **Faster R-CNN** (`fasterrcnn_resnet50_fpn_v2`) object detector on a sample image.

```python
import torch
import torchvision
from torchvision.models.detection import fasterrcnn_resnet50_fpn_v2, FasterRCNN_ResNet50_FPN_V2_Weights

boxes = torch.tensor([[10., 10., 50., 50.], [14., 14., 54., 54.], [100., 100., 140., 140.]])
scores = torch.tensor([0.95, 0.82, 0.88])
keep = torchvision.ops.nms(boxes, scores, iou_threshold=0.5)
print("Kept box indices after NMS:", keep.tolist())
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
weights = FasterRCNN_ResNet50_FPN_V2_Weights.DEFAULT
detector = fasterrcnn_resnet50_fpn_v2(weights=weights, box_score_thresh=0.7).to(device).eval()

# Run inference on a synthetic tensor (replace with PIL Image -> weights.transforms())
sample = torch.rand(3, 320, 320, device=device)
with torch.no_grad():
    preds = detector([sample])
print("Detections returned:", len(preds[0]["boxes"]))
```
