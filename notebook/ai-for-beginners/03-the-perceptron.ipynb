In this notebook, you will:
1. Implement a vectorized NumPy **Binary Perceptron** and visualize its decision boundary.
2. Complete the **Multi-Class MNIST Perceptron Lab** using a One-vs-Rest weight matrix.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
pos = np.random.randn(50, 2) + np.array([2.0, 2.0])
neg = np.random.randn(50, 2) + np.array([-2.0, -2.0])
X = np.vstack([pos, neg])
X_bias = np.hstack([X, np.ones((X.shape[0], 1))])
y = np.array([1]*50 + [-1]*50)
```

```python
def train_perceptron(X_b, y_true, epochs=15, lr=0.1):
    w = np.zeros(X_b.shape[1])
    for ep in range(epochs):
        errors = 0
        for xi, yi in zip(X_b, y_true):
            pred = 1 if np.dot(xi, w) >= 0 else -1
            if pred != yi:
                w += lr * yi * xi
                errors += 1
        if errors == 0:
            print(f"Converged at epoch {ep + 1}")
            break
    return w

w = train_perceptron(X_bias, y)
print("Learned weights:", w)
```

## Lab: Multi-Class Digit Perceptron on Scikit-Learn Digits (8x8 MNIST)

We train 10 linear perceptrons (one per digit `0`–`9`) and evaluate accuracy on a held-out test split.

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split

digits = load_digits()
X_d = digits.data / 16.0
X_d = np.hstack([X_d, np.ones((X_d.shape[0], 1))])
y_d = digits.target

X_tr, X_te, y_tr, y_te = train_test_split(X_d, y_d, test_size=0.2, random_state=42)
W = np.zeros((10, X_tr.shape[1]))
lr = 0.05

for epoch in range(25):
    for xi, yi in zip(X_tr, y_tr):
        scores = W @ xi
        pred = np.argmax(scores)
        if pred != yi:
            W[yi] += lr * xi
            W[pred] -= lr * xi

preds = np.argmax(X_te @ W.T, axis=1)
print(f"Multi-Class Perceptron Test Accuracy: {(preds == y_te).mean()*100:.2f}%")
```
