# Lesson 14: Semantic Word Embeddings (`nn.Embedding` & Word2Vec)

In this notebook, you will:
1. Build a PyTorch `nn.EmbeddingBag` text classifier.
2. Train/inspect semantic word vectors with `gensim` Word2Vec and compute cosine analogies.

```python
import torch
import torch.nn as nn

class EmbeddingBagClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, num_classes):
        super().__init__()
        self.emb = nn.EmbeddingBag(vocab_size, embed_dim, mode="mean")
        self.fc = nn.Linear(embed_dim, num_classes)
    def forward(self, text_ids, offsets):
        return self.fc(self.emb(text_ids, offsets))

model = EmbeddingBagClassifier(vocab_size=1000, embed_dim=32, num_classes=4)
sample_ids = torch.tensor([1, 42, 7, 19, 88])
offsets = torch.tensor([0, 3])
print("Logits shape for 2 documents:", tuple(model(sample_ids, offsets).shape))
```

```python
!pip install -q gensim
from gensim.models import Word2Vec

sentences = [
    "neural networks learn features from data".split(),
    "deep learning models use pytorch tensors".split(),
    "convolutional neural networks excel at computer vision".split(),
    "recurrent networks and transformers process text sequences".split(),
] * 20

w2v = Word2Vec(sentences, vector_size=16, window=3, min_count=1, epochs=50, seed=42)
print("Most similar to 'networks':", w2v.wv.most_similar("networks", topn=3))
```
