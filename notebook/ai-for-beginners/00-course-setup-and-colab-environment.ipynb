## 1. Quick Tensor Smoke Test

Let's create two matrices and multiply them on the active device (`cuda` if you enabled T4 GPU, otherwise `cpu`).

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
a = torch.randn(512, 512, device=device)
b = torch.randn(512, 512, device=device)
c = a @ b
print(f"Matrix multiplication succeeded on {device}. Output shape: {tuple(c.shape)}")
```
