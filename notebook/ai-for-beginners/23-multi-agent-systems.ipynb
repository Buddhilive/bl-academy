# Lesson 23: Multi-Agent Systems & Emergent Behavior

In this notebook, you will:
1. Simulate **2D Boids Flocking** (Separation, Alignment, Cohesion) in Python.
2. Visualize emergent group trajectories with Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
N = 25
pos = np.random.rand(N, 2) * 100
vel = (np.random.rand(N, 2) - 0.5) * 4

for step in range(40):
    new_vel = vel.copy()
    for i in range(N):
        dists = np.linalg.norm(pos - pos[i], axis=1)
        nbrs = (dists > 0) & (dists < 25.0)
        if nbrs.any():
            # Alignment + Cohesion
            new_vel[i] += 0.05 * (vel[nbrs].mean(axis=0) - vel[i])
            new_vel[i] += 0.01 * (pos[nbrs].mean(axis=0) - pos[i])
        close = (dists > 0) & (dists < 5.0)
        if close.any():
            # Separation
            new_vel[i] -= 0.1 * (pos[close] - pos[i]).mean(axis=0)
    vel = new_vel
    pos = (pos + vel) % 100

plt.figure(figsize=(5, 5))
plt.quiver(pos[:, 0], pos[:, 1], vel[:, 0], vel[:, 1])
plt.title("Emergent Boids Flocking After 40 Steps")
plt.show()
```

```python
# Lotka-Volterra Predator-Prey Multi-Agent Population Dynamics
prey, pred = 40.0, 9.0
history = []
for _ in range(100):
    d_prey = (0.1 * prey - 0.02 * prey * pred) * 0.1
    d_pred = (0.01 * prey * pred - 0.1 * pred) * 0.1
    prey, pred = prey + d_prey, pred + d_pred
    history.append((prey, pred))
print(f"Final Predator-Prey state -> Prey: {prey:.1f}, Predators: {pred:.1f}")
```
