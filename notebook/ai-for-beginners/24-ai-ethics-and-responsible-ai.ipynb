# Lesson 24: AI Ethics, Fairness & Responsible AI (`Fairlearn` & `InterpretML`)

In this notebook, you will:
1. Measure **Demographic Parity** and **Equalized Odds** using open-source `fairlearn`.
2. Mitigate classifier bias using `ExponentiatedGradient` with a `DemographicParity` constraint.

```python
!pip install -q fairlearn scikit-learn
import numpy as np
from sklearn.linear_model import LogisticRegression
from fairlearn.metrics import MetricFrame, selection_rate, demographic_parity_difference
from fairlearn.reductions import ExponentiatedGradient, DemographicParity

np.random.seed(42)
n = 400
group = np.random.binomial(1, 0.5, size=n)
skill = np.random.normal(0, 1, size=n)
# Historical label bias favoring group==1
y = ((skill + 0.8 * group) > 0.2).astype(int)
X = np.column_stack([skill, group])

clf = LogisticRegression().fit(X, y)
preds = clf.predict(X)
print(f"Unmitigated Demographic Parity Diff: {demographic_parity_difference(y, preds, sensitive_features=group):.3f}")
```

```python
# Mitigate bias with Fairlearn ExponentiatedGradient
mitigator = ExponentiatedGradient(LogisticRegression(), DemographicParity())
mitigator.fit(X, y, sensitive_features=group)
mit_preds = mitigator.predict(X)
print(f"Mitigated Demographic Parity Diff:   {demographic_parity_difference(y, mit_preds, sensitive_features=group):.3f}")
```
